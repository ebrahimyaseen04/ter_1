```python
# ==========================================
# Step 1: Mount Google Drive
# ==========================================
from google.colab import drive

drive.mount('/content/drive')

# ==========================================
# Step 2 & 3: Student Form and Update Logic
# ==========================================
# @title Student Management Form { display-mode: "form" }
import os

file_path = '/content/drive/MyDrive/students.txt'

# 1. Read student data from file
students_data = {}
if os.path.exists(file_path):
  with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
      line_clean = line.strip()
      if not line_clean:
        continue
      parts = [p.strip() for p in line_clean.split(',')]
      if len(parts) >= 5:
        full_name = f'{parts[0]} {parts[1]}'
        students_data[full_name] = {
            'first_name': parts[0],
            'last_name': parts[1],
            'email': parts[2],
            'courses': parts[3],
            'link': parts[4],
        }

student_options = (
    list(students_data.keys()) if students_data else ['No Students Found']
)

# 2. Form Fields (Colab GUI)
# @markdown ### Select Student:
Selected_Student = 'Yossi Cohen'  # @param ['Yossi Cohen', 'Dana Levi'] {allow-input: true}

# @markdown ### Favorite TV Show / Program:
Favorite_Program = 'Breaking Bad'  # @param {type:"string"}

# @markdown ### Click to Update File:
Update_Data = False  # @param {type:"boolean"}

# 3. Display Selected Student Details
if Selected_Student in students_data:
  info = students_data[Selected_Student]
  print(f'=== Student Details: {Selected_Student} ===')
  print(f"First Name: {info['first_name']}")
  print(f"Last Name:  {info['last_name']}")
  print(f"Email:      {info['email']}")
  print(f"Courses:    {info['courses']}")
  print(f"Link:       {info['link']}")
else:
  print('Selected student is not in the list.')

# 4. Append updated row if checkbox is ticked
if Update_Data and Selected_Student in students_data:
  info = students_data[Selected_Student]
  new_entry = f"\n{info['first_name']}, {info['last_name']}, {info['email']}, {info['courses']}, {info['link']}, {Favorite_Program}"
  with open(file_path, 'a', encoding='utf-8') as f:
    f.write(new_entry)
  print('\n✅ Success: Updated entry added to students.txt!')
```
